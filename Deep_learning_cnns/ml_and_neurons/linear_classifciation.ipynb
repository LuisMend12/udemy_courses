```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    data.data, data.target, test_size=0.33)
```

```python
N, D = X_train.shape
```

```python
N, D
```

```python
from sklearn.preprocessing import StandardScaler #scaling the data, that is fitting the data at a reasonable scale

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train, X_test
```

```python
model = tf.keras.models.Sequential([
    tf.keras.layers.Input(shape=(D,)),
    tf.keras.layers.Dense(1, activation='sigmoid')
])
```

```python
#alterntive
# model = tf.keras.models.Sequential()
# model.add(tf.keras.layers.Dense(1, input_shape=(D, ), activation='sigmoid'))
```

```python
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])
```

```python
r = model.fit(X_train, y_train, validation_data=(X_test, y_test), epochs=100)
```

```python
print("Train scores:", model.evaluate(X_train, y_train))
print("Test scores:", model.evaluate(X_test, y_test))
```

```python
import matplotlib.pyplot as plt

plt.plot(r.history['loss'], label='loss')
plt.plot(r.history['val_loss'], label='val_loss')
plt.legend()
```

```python
plt.plot(r.history['accuracy'], label='accuracy')
plt.plot(r.history['val_accuracy'], label='val_accuracy')
plt.legend()
```
